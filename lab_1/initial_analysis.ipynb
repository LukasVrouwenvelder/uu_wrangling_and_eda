### 6) What is the average amount each university receives? Median amount? If they are different why?

By Institution 1) *see plot*
- Median and mean are different because the data is assymetrical. In this case, 'mean of all donations' > 'median of all donations', which indicates that there are large donations contributing to a right-skewed distribution. 



Aggregate 2) 
- Mean accross all Institutions = $588,233
- Median accross all Institutions = $94,615
    - This comparison also illustrates the right skew of the distribution. 



Mean & Median across each *INDIVIDUAL* Institution's donations

```python
# group by insitution, take the average of all their donations, sort by avg amount, descending
df.groupby('Institution Name')['Foreign Gift Amount'].mean().sort_values(ascending=False)
```

```python
# group by insitution, take the median of all their donations, sort by avg amount, descending

df.groupby('Institution Name')['Foreign Gift Amount'].median().sort_values(ascending=False)
```

```python
# donationPlt = px.bar(df.groupby('Institution Name')['Foreign Gift Amount'].mean().sort_values(ascending=False).head(20))



# # add lines for median and mean of all donations, regardless of country or institution
# donationPlt.add_hline(y=df['Foreign Gift Amount'].mean(), line_color="black", annotation_text="Mean of all donations", annotation_position="bottom right")
# donationPlt.add_hline(y=df['Foreign Gift Amount'].median(), line_color="red", annotation_text="Median of all donations", annotation_position="top right")
```

```python
grouped = (
    df.groupby('Institution Name')['Foreign Gift Amount']
    .mean()
    .sort_values(ascending=False)
    .head(20)
)

donationPlt = px.bar(
    x=grouped.values,
    y=grouped.index,
    orientation='h',
    labels={'x': 'Average Foreign Gift Amount', 'y': 'Institution Name'},
)
fGA = "${:,.0f}".format(df['Foreign Gift Amount'].mean())
mGA = "${:,.0f}".format(df['Foreign Gift Amount'].median())

donationPlt.update_yaxes(autorange="reversed")
donationPlt.add_vline(x=df['Foreign Gift Amount'].mean(), line = dict( color ="black", width = 3), annotation_text= f"Mean of all donations {fGA}", annotation_position="bottom right", annotation_yshift=-30,annotation_xshift = -10)
donationPlt.add_vline(x=df['Foreign Gift Amount'].median(), line = dict(color="red", width = 3) , annotation_text= f"Median of all donations {mGA}", annotation_position="top right", annotation_yshift=20, annotation_xshift = -10, annotation_font_color="red")
donationPlt.show()
```

### 8) Which are the top giftors to US academic institutions?

- The top-5 giftors are: 
    - Qatar: 2,706,240,869
    - England: 1,464,906,771
    - China: 1,237,952,112
    - Saudi Arabia: 1,065,205,930
    - Bermuda: 899,593,972

```python
# df.head()

#giftor = 'Giftor Name'
giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 25

flows = (
    df.groupby([giftor, 
                recipi])
      [flow]
      .sum()
      .nlargest(N)
      .reset_index()
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```
